```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print(df["Order_Date"].dtype)
```

```python
# KPI Analysis

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order_ID"].nunique()
total_quantity = df["Quantity"].sum()
average_order_value = df["Sales"].mean()
profit_margin = (total_profit / total_sales) * 100

print("===== SALES KPIs =====")
print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Quantity Sold:", total_quantity)
print("Average Order Value:", round(average_order_value, 2))
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
# KPI Summary Table

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Sales",
        "Total Profit",
        "Total Orders",
        "Total Quantity Sold",
        "Average Order Value",
        "Profit Margin"
    ],
    "Value": [
        round(total_sales, 2),
        round(total_profit, 2),
        total_orders,
        total_quantity,
        round(average_order_value, 2),
        round(profit_margin, 2)
    ]
})

kpi_summary
```

```python
# Product Sales Analysis

product_sales = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Product:")
display(product_sales)
```

```python
# Sales by Product Visualization

plt.figure(figsize=(10, 5))

product_sales.plot(kind="bar")

plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Top Selling Product

top_product = product_sales.idxmax()
top_product_sales = product_sales.max()

print("Top Selling Product:", top_product)
print("Sales:", round(top_product_sales, 2))
```

```python
# Regional Sales Analysis

region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Region:")
display(region_sales)
```

```python
# Sales by Region Visualization

plt.figure(figsize=(10, 5))

region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Best Performing Region

best_region = region_sales.idxmax()
best_region_sales = region_sales.max()

print("Best Performing Region:", best_region)
print("Sales:", round(best_region_sales, 2))
```

```python
# Category Sales Analysis

category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Category:")
display(category_sales)
```

```python
# Category Profit Analysis

category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print("Profit by Category:")
display(category_profit)
```

```python
# Profit by Category Visualization

plt.figure(figsize=(10, 5))

category_profit.plot(kind="bar")

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Most Profitable Category

best_category = category_profit.idxmax()
best_category_profit = category_profit.max()

print("Most Profitable Category:", best_category)
print("Profit:", round(best_category_profit, 2))
```

```python
# Monthly Sales Analysis

df["Month"] = df["Order_Date"].dt.to_period("M")

monthly_sales = (
    df.groupby("Month")["Sales"]
    .sum()
)

print("Monthly Sales:")
display(monthly_sales)
```

```python
# Monthly Sales Trend Visualization

plt.figure(figsize=(10, 5))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
# Best Sales Month

best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()

print("Best Sales Month:", best_month)
print("Sales:", round(best_month_sales, 2))
```

```python
# Lowest Sales Month

lowest_month = monthly_sales.idxmin()
lowest_month_sales = monthly_sales.min()

print("Lowest Sales Month:", lowest_month)
print("Sales:", round(lowest_month_sales, 2))
```

```python
# Profit Margin Analysis

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Overall Profit Margin:", round(profit_margin, 2), "%")
```

```python
# Profit by Product

product_profit = (
    df.groupby("Product")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print("Profit by Product:")
display(product_profit)
```

```python
# Profit by Product Visualization

plt.figure(figsize=(10, 5))

product_profit.plot(kind="bar")

plt.title("Profit by Product")
plt.xlabel("Product")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Most Profitable Product

most_profitable_product = product_profit.idxmax()
highest_product_profit = product_profit.max()

print("Most Profitable Product:", most_profitable_product)
print("Profit:", round(highest_product_profit, 2))
```

```python
# Business Insights Summary

print("===== BUSINESS INSIGHTS =====")

print("\n1. Top Selling Product:")
print(top_product)

print("\n2. Best Performing Region:")
print(best_region)

print("\n3. Most Profitable Category:")
print(best_category)

print("\n4. Most Profitable Product:")
print(most_profitable_product)

print("\n5. Best Sales Month:")
print(best_month)

print("\n6. Lowest Sales Month:")
print(lowest_month)

print("\n7. Overall Profit Margin:")
print(round(profit_margin, 2), "%")
```
